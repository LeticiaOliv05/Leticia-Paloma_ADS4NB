```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import nltk
import spacy

from nltk.corpus import stopwords
from gensim.models import FastText
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# PREPARAÇÃO DO AMBIENTE

nltk.download("stopwords", quiet=True)

stop_words_pt = set(stopwords.words("portuguese"))

# Carrega o modelo de português do spaCy
nlp = spacy.load("pt_core_news_sm")

print("Ambiente preparado com sucesso.")


# FUNÇÃO DE PRÉ-PROCESSAMENTO


def limpar_e_lemmatizar(texto):
    """
    Realiza o pré-processamento de uma mensagem:

    1. Converte para minúsculas
    2. Remove caracteres especiais e números
    3. Tokeniza utilizando spaCy
    4. Remove stopwords
    5. Realiza lemmatization
    6. Remove tokens muito curtos
    7. Retorna o texto normalizado
    """

    # 1. Converter para minúsculas
    texto_limpo = texto.lower()

    # 2. Remover caracteres especiais e números
    texto_limpo = re.sub(
        r'[^a-záàâãéèêíïóôõöúçñ\s]',
        '',
        texto_limpo
    )

    # 3. Processar o texto com spaCy
    doc = nlp(texto_limpo)

    # Lista para armazenar os tokens finais
    tokens_filtrados = []

    # 4, 5 e 6. Filtrar e lematizar
    for token in doc:

        # Ignorar espaços
        if token.is_space:
            continue

        # Ignorar stopwords
        if token.text in stop_words_pt:
            continue

        # Ignorar tokens muito pequenos
        if len(token.text) <= 1:
            continue

        # Adicionar o lema
        tokens_filtrados.append(token.lemma_)

    # 7. Reconstruir a frase
    return " ".join(tokens_filtrados)



# FUNÇÃO DE DIAGNÓSTICO


def analisar_preprocessamento(texto):
    """
    Apresenta um diagnóstico detalhado do pré-processamento.
    """

    # Texto em minúsculas
    texto_minusculo = texto.lower()

    # Remoção de caracteres especiais e números
    texto_limpo = re.sub(
        r'[^a-záàâãéèêíïóôõöúçñ\s]',
        '',
        texto_minusculo
    )

    # Tokenização inicial
    doc = nlp(texto_limpo)

    tokens_antes = [
        token.text
        for token in doc
        if not token.is_space
    ]

    # Processamento final
    tokens_finais = []

    for token in doc:

        if token.is_space:
            continue

        if token.text in stop_words_pt:
            continue

        if len(token.text) <= 1:
            continue

        tokens_finais.append(token.lemma_)

    # Texto normalizado
    texto_normalizado = " ".join(tokens_finais)

    # Tokens removidos
    tokens_removidos = [
        token
        for token in tokens_antes
        if token not in tokens_finais
        and token not in [nlp(t)[0].lemma_ for t in tokens_finais]
    ]

    # Exibição
    print("=" * 60)
    print("DIAGNÓSTICO DO PRÉ-PROCESSAMENTO")
    print("=" * 60)

    print("\nTexto original:")
    print(texto)

    print("\nTexto normalizado:")
    print(texto_normalizado)

    print("\nQuantidade de caracteres:")
    print(len(texto))

    print("\nQuantidade de tokens antes:")
    print(len(tokens_antes))

    print("\nQuantidade de tokens depois:")
    print(len(tokens_finais))

    print("\nTokens removidos:")
    print(tokens_removidos)

    print("\nTokens finais:")
    print(tokens_finais)

    print("=" * 60)



# TESTE 1


frase_teste = (
    "Olá!!! EU gostaria de saber se vocês estão "
    "DEVOLVENDO as mesas que foram compradas ontem."
)

resultado = limpar_e_lemmatizar(frase_teste)

print("\nFrase original:")
print(frase_teste)

print("\nFrase processada:")
print(resultado)


# TESTE 2 - DIAGNÓSTICO


analisar_preprocessamento(
    "MEU sofá!!! chegou quebrado e quero DEVOLVER!!!"
)
```

```python

# FASTTEXT + MEAN POOLING


import numpy as np
import pandas as pd
from gensim.models import FastText


# 1. CARREGAR O DATASET


df = pd.read_csv("sac_moveis_ac2_treino.csv")

print("Dataset carregado com sucesso!")

print("\nDimensões do dataset:")
print(df.shape)

print("\nColunas disponíveis:")
print(df.columns.tolist())

print("\nPrimeiras mensagens:")
print(df.head())



# 2. FUNÇÃO DO EXERCÍCIO 1


df["mensagem_normalizada"] = df["mensagem"].apply(
    limpar_e_lemmatizar
)

print("\nMensagens normalizadas:")
print(
    df[["mensagem", "mensagem_normalizada"]].head()
)


# 3. TRANSFORMAR AS MENSAGENS EM LISTAS DE TOKENS


corpus_tokenizado = (
    df["mensagem_normalizada"]
    .apply(lambda texto: texto.split())
    .tolist()
)

print("\nCorpus tokenizado:")
print(corpus_tokenizado[:3])


# 4. CONSTRUIR O MODELO FASTTEXT


modelo_fasttext = FastText(
    sentences=corpus_tokenizado,
    vector_size=50,
    window=3,
    min_count=1,
    workers=4,
    sg=1
)

print("\nModelo FastText treinado com sucesso!")

print(
    "Tamanho do vocabulário:",
    len(modelo_fasttext.wv)
)

print(
    "Dimensão dos vetores:",
    modelo_fasttext.vector_size
)



# 5. FUNÇÃO DE MEAN POOLING


def obter_vetor_frase(frase, modelo):
    """
    Transforma uma frase em um vetor denso utilizando
    FastText + Mean Pooling.
    """

    # Separar a frase em palavras
    palavras = frase.split()

    # Lista para armazenar os vetores
    vetores = []

    for palavra in palavras:

        # TODO 1:
        # Obter o vetor da palavra no modelo FastText
        vetor = modelo.wv[palavra]

        # TODO 2:
        # Adicionar o vetor à lista
        vetores.append(vetor)

    # TODO 3:
    # Verificar se nenhum vetor foi encontrado
    if len(vetores) == 0:
        return np.zeros(modelo.vector_size)

    # TODO 4:
    # Calcular o Mean Pooling
    vetor_medio = np.mean(vetores, axis=0)

    return vetor_medio


# 6. TESTAR A FUNÇÃO


frase_teste = "quero devolver meu sofá"

# Aplicar o pré-processamento
frase_processada = limpar_e_lemmatizar(
    frase_teste
)

# Gerar vetor da frase
vetor_frase = obter_vetor_frase(
    frase_processada,
    modelo_fasttext
)

print("\n" + "=" * 60)
print("TESTE DO MEAN POOLING")
print("=" * 60)

print("\nFrase original:")
print(frase_teste)

print("\nFrase processada:")
print(frase_processada)

print("\nVetor da frase:")
print(vetor_frase)

print("\nDimensão do vetor:")
print(vetor_frase.shape)


# 7. GERAR VETORES PARA TODAS AS MENSAGENS


df["vetor"] = df["mensagem_normalizada"].apply(
    lambda frase: obter_vetor_frase(
        frase,
        modelo_fasttext
    )
)

print("\nVetores gerados para todas as mensagens!")



# 8. CRIAR MATRIZ DE EMBEDDINGS


X_embeddings = np.vstack(
    df["vetor"].values
)

print("\n" + "=" * 60)
print("MATRIZ DE EMBEDDINGS")
print("=" * 60)

print("\nFormato da matriz:")
print(X_embeddings.shape)

print("\nDimensão de cada vetor:")
print(X_embeddings.shape[1])


#
# 9. RESUMO FINAL


print("\n" + "=" * 60)
print("RESUMO DO EXERCÍCIO 2")
print("=" * 60)

print("\nQuantidade de mensagens:")
print(len(df))

print("\nTamanho do vocabulário:")
print(len(modelo_fasttext.wv))

print("\nDimensão dos embeddings:")
print(modelo_fasttext.vector_size)

print("\nFormato da matriz final:")
print(X_embeddings.shape)

print("\nExercício 2 concluído com sucesso!")
```

```python

# CLASSIFICADOR DE INTENÇÕES COM FALLBACK

import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)



# 1. SEPARAR OS DADOS


# Vetores gerados no Exercício 2
X_vetores = X_embeddings

# Intenções do dataset
y = df["intencao"]


X_train, X_test, y_train, y_test = train_test_split(
    X_vetores,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("=" * 60)
print("DIVISÃO DOS DADOS")
print("=" * 60)

print("\nDados de treinamento:")
print(X_train.shape)

print("\nDados de teste:")
print(X_test.shape)

print("\nTreinamento:")
print(len(y_train))

print("\nTeste:")
print(len(y_test))



# 2. TREINAR A REGRESSÃO LOGÍSTICA


modelo = LogisticRegression(
    max_iter=1000
)

modelo.fit(
    X_train,
    y_train
)

print("\n" + "=" * 60)
print("MODELO TREINADO")
print("=" * 60)

print("\nRegressão Logística treinada com sucesso!")


# 3. REALIZAR AS PREVISÕES


y_pred = modelo.predict(
    X_test
)


# 4. AVALIAR O MODELO


accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n" + "=" * 60)
print("AVALIAÇÃO DO MODELO")
print("=" * 60)

print("\nAccuracy:")
print(accuracy)

print("\nRelatório de classificação:")
print(
    classification_report(
        y_test,
        y_pred
    )
)


# 5. MATRIZ DE CONFUSÃO


matriz = confusion_matrix(
    y_test,
    y_pred
)

print("\n" + "=" * 60)
print("MATRIZ DE CONFUSÃO")
print("=" * 60)

print(matriz)


# 6. FUNÇÃO DE CLASSIFICAÇÃO COM FALLBACK


def classificar_mensagem(
    mensagem,
    modelo,
    modelo_embedding,
    limiar=0.50
):
    """
    Classifica uma mensagem utilizando:

    1. Pré-processamento
    2. FastText + Mean Pooling
    3. Regressão Logística
    4. Confiança mínima de 50%

    Caso a confiança seja menor que o limiar,
    retorna FALLBACK_HUMANO.
    """


    # 1. Pré-processar a mensagem


    mensagem_processada = limpar_e_lemmatizar(
        mensagem
    )


    # 2. Transformar a mensagem em vetor


    vetor = obter_vetor_frase(
        mensagem_processada,
        modelo_embedding
    )


    # 3. Ajustar formato do vetor


    vetor = vetor.reshape(1, -1)


    # 4. Obter probabilidades


    probabilidades = modelo.predict_proba(
        vetor
    )[0]

    # 5. Identificar maior probabilidade


    indice_maior = np.argmax(
        probabilidades
    )

    intencao = modelo.classes_[
        indice_maior
    ]

    confianca = probabilidades[
        indice_maior
    ]


    # 6. Aplicar o Fallback


    if confianca >= limiar:

        resultado = intencao

    else:

        resultado = "FALLBACK_HUMANO"

    return resultado, confianca



# 7. TESTAR O CHATBOT


testes = [
    "quero devolver meu sofá",
    "como faço para realizar a devolução?",
    "cadê meu pedido?",
    "meu pedido nao chego",
    "qual é a previsão do tempo?"
]


print("\n" + "=" * 60)
print("TESTE DO CHATBOT")
print("=" * 60)


for mensagem in testes:

    resultado, confianca = classificar_mensagem(
        mensagem,
        modelo,
        modelo_fasttext,
        limiar=0.50
    )

    print("\nMensagem:")
    print(mensagem)

    print("Intenção:")
    print(resultado)

    print("Confiança:")
    print(f"{confianca:.2%}")

    print("-" * 60)
```

```python
# LABORATÓRIO COMPARATIVO:
# REGRESSÃO LOGÍSTICA × KNN


import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


# 1. UTILIZAR OS DADOS SEPARADOS NO EXERCÍCIO 3


X_train_vec = X_train
X_test_vec = X_test



# 2. CRIAR OS MODELOS


modelo_logistico = LogisticRegression(
    max_iter=1000
)

modelo_knn = KNeighborsClassifier(
    n_neighbors=3
)


# 3. TREINAR OS DOIS MODELOS


modelo_logistico.fit(
    X_train_vec,
    y_train
)

modelo_knn.fit(
    X_train_vec,
    y_train
)

print("=" * 60)
print("MODELOS TREINADOS")
print("=" * 60)

print("\nRegressão Logística: OK")
print("KNN: OK")


# 4. FAZER AS PREVISÕES NO CONJUNTO DE TESTE


y_pred_logistico = modelo_logistico.predict(
    X_test_vec
)

y_pred_knn = modelo_knn.predict(
    X_test_vec
)


# 5. CALCULAR AS MÉTRICAS

# REGRESSÃO LOGÍSTICA


accuracy_logistico = accuracy_score(
    y_test,
    y_pred_logistico
)

precision_logistico = precision_score(
    y_test,
    y_pred_logistico,
    average="weighted",
    zero_division=0
)

recall_logistico = recall_score(
    y_test,
    y_pred_logistico,
    average="weighted",
    zero_division=0
)

f1_logistico = f1_score(
    y_test,
    y_pred_logistico,
    average="weighted",
    zero_division=0
)



# KNN


accuracy_knn = accuracy_score(
    y_test,
    y_pred_knn
)

precision_knn = precision_score(
    y_test,
    y_pred_knn,
    average="weighted",
    zero_division=0
)

recall_knn = recall_score(
    y_test,
    y_pred_knn,
    average="weighted",
    zero_division=0
)

f1_knn = f1_score(
    y_test,
    y_pred_knn,
    average="weighted",
    zero_division=0
)


# 6. CRIAR A TABELA COMPARATIVA


resultados = pd.DataFrame({
    "Modelo": [
        "Regressão Logística",
        "KNN"
    ],

    "Accuracy": [
        accuracy_logistico,
        accuracy_knn
    ],

    "Precision": [
        precision_logistico,
        precision_knn
    ],

    "Recall": [
        recall_logistico,
        recall_knn
    ],

    "F1": [
        f1_logistico,
        f1_knn
    ]
})



# 7. FORMATAR OS RESULTADOS EM PORCENTAGEM


resultados_formatados = resultados.copy()

for coluna in [
    "Accuracy",
    "Precision",
    "Recall",
    "F1"
]:

    resultados_formatados[coluna] = (
        resultados_formatados[coluna] * 100
    ).round(2).astype(str) + "%"



# 8. EXIBIR A TABELA FINAL


print("\n" + "=" * 60)
print("COMPARAÇÃO DOS MODELOS")
print("=" * 60)

print(
    resultados_formatados.to_string(
        index=False
    )
)


# 9. IDENTIFICAR O MELHOR MODELO


melhor_modelo = resultados.loc[
    resultados["F1"].idxmax(),
    "Modelo"
]

melhor_f1 = resultados["F1"].max()

print("\n" + "=" * 60)
print("RECOMENDAÇÃO TÉCNICA")
print("=" * 60)

print(
    f"\nModelo recomendado: {melhor_modelo}"
)

print(
    f"F1-Score: {melhor_f1:.2%}"
)
```
